### Ejercicio 3 — Ridge reparte, Lasso elige: el caso $n = p = 2$

Es sabido que la regresión Ridge tiende a devolver coeficientes similares para variables correlacionadas, mientras que Lasso puede dar coeficientes muy diferentes para variables correlacionadas. Vamos a explorar esta propiedad en un escenario muy simple.  

Supongamos que $ n = 2, p = 2, x_{11} = x_{12}, x_{21} = x_{22} $. Además, supongamos que $ y_1 + y_2 = 0 $ y $ x_{11} + x_{21} = 0 $ y $ x_{12} + x_{22} = 0 $, de modo que el estimador de la intersección en un modelo de mínimos cuadrados, regresión Ridge o Lasso es cero: $\hat{\beta}_0 = 0$.

1. Escribir el problema de optimización Ridge en este escenario.  
1. Ver que, en este escenario, las estimaciones de los coeficientes Ridge satisfacen $\hat{\beta}_1 = \hat{\beta}_2$.  

1. Escriba el problema de optimización Lasso en este escenario.  

1. Ver que, en este escenario, los coeficientes de Lasso $\hat{\beta}_1$ y $\hat{\beta}_2$ no son únicos; en otras palabras, hay muchas soluciones posibles para el problema de optimización anterior. Describa esas soluciones.


**Mi respuesta:**

**Preparación.** Llamemos $a := x_{11} = x_{12}$. Como $x_{11} + x_{21} = 0$, se tiene $x_{21} = x_{22} = -a$; y como $y_1 + y_2 = 0$, $y_2 = -y_1$. Suponemos $a \neq 0$ (si no, $\mathbf{X} = 0$ y todo es trivial). Con $\hat\beta_0 = 0$ las predicciones son

$$\hat y_1 = a\beta_1 + a\beta_2 = a(\beta_1+\beta_2), \qquad \hat y_2 = -a(\beta_1 + \beta_2)$$

y los dos residuos valen lo mismo en módulo: $y_2 - \hat y_2 = -y_1 + a(\beta_1+\beta_2) = -\big(y_1 - a(\beta_1+\beta_2)\big)$. Entonces el RSS es

$$\text{RSS}(\beta) = 2\big(y_1 - a(\beta_1+\beta_2)\big)^2$$

— igual que en el Ejercicio 3 de la clase 4, **depende solo de la suma** $s = \beta_1 + \beta_2$.

<!-- 3.1 — Escribir el problema de optimización Ridge en este escenario -->
**3.1 — Problema de Ridge.**

$$\min_{\beta_1,\beta_2}\ \big(y_1 - a\beta_1 - a\beta_2\big)^2 + \big(y_2 + a\beta_1 + a\beta_2\big)^2 + \lambda\left(\beta_1^2 + \beta_2^2\right)$$

que con las simplificaciones de arriba queda

$$\min_{\beta_1,\beta_2}\ 2\big(y_1 - a(\beta_1+\beta_2)\big)^2 + \lambda\left(\beta_1^2 + \beta_2^2\right)$$

<!-- 3.2 — Ver que, en este escenario, las estimaciones de los coeficientes Ridge satisfacen $\hat{\beta}_1 =… -->
**3.2 — Ridge da $\hat\beta_1 = \hat\beta_2$.** El objetivo es diferenciable; igualamos a cero las derivadas parciales:

$$\frac{\partial}{\partial\beta_1}: \ -4a\big(y_1 - a(\beta_1+\beta_2)\big) + 2\lambda\beta_1 = 0$$
$$\frac{\partial}{\partial\beta_2}: \ -4a\big(y_1 - a(\beta_1+\beta_2)\big) + 2\lambda\beta_2 = 0$$

El primer término es **idéntico** en las dos ecuaciones (porque el RSS solo ve la suma). Restándolas:

$$2\lambda(\hat\beta_1 - \hat\beta_2) = 0 \ \Longrightarrow\ \hat\beta_1 = \hat\beta_2 \quad (\lambda > 0)$$

Y es la única solución: para $\lambda > 0$ el objetivo es estrictamente convexo (el término $\lambda\|\beta\|^2$ lo es), así que el punto crítico es el mínimo global y es único. Reemplazando $\beta_1 = \beta_2 = \beta$ en cualquiera de las ecuaciones: $-4ay_1 + 8a^2\beta + 2\lambda\beta = 0$, o sea

$$\hat\beta_1^{ridge} = \hat\beta_2^{ridge} = \frac{2ay_1}{4a^2 + \lambda}$$

La intuición es la del ejercicio de la clase 4: para una suma fija, el RSS es indiferente al reparto y $\beta_1^2 + \beta_2^2$ prefiere el reparto parejo.

<!-- 3.3 — Escriba el problema de optimización Lasso en este escenario -->
**3.3 — Problema de Lasso.**

$$\min_{\beta_1,\beta_2}\ \big(y_1 - a\beta_1 - a\beta_2\big)^2 + \big(y_2 + a\beta_1 + a\beta_2\big)^2 + \lambda\left(|\beta_1| + |\beta_2|\right)$$

es decir

$$\min_{\beta_1,\beta_2}\ 2\big(y_1 - a(\beta_1+\beta_2)\big)^2 + \lambda\left(|\beta_1| + |\beta_2|\right)$$

<!-- 3.4 — Ver que, en este escenario, los coeficientes de Lasso $\hat{\beta}_1$ y $\hat{\beta}_2$ no son… -->
**3.4 — Lasso no tiene solución única.** La clave es que la penalización $L_1$, a diferencia de la $L_2$, **no tiene preferencia por ningún reparto** entre dos números del mismo signo. Por la desigualdad triangular,

$$|\beta_1| + |\beta_2| \ \ge\ |\beta_1 + \beta_2| = |s|$$

con **igualdad si y solo si $\beta_1$ y $\beta_2$ tienen el mismo signo** (o alguno es $0$), es decir $\beta_1\beta_2 \ge 0$. Así que, para cualquier $\beta$,

$$\text{objetivo}(\beta) \ \ge\ 2(y_1 - as)^2 + \lambda|s| =: f(s)$$

y la igualdad se alcanza en todos los pares con suma $s$ y $\beta_1\beta_2 \ge 0$. El problema en dos variables se reduce a minimizar $f$ en **una** variable. $f$ es un Lasso unidimensional; su solución es *soft-thresholding* (para $s > 0$: $f'(s) = -4a(y_1 - as) + \lambda = 0 \Rightarrow s = \frac{4ay_1 - \lambda}{4a^2}$, y análogo para $s < 0$):

$$\hat s = \frac{\operatorname{sign}(ay_1)\,\big(|ay_1| - \lambda/4\big)_+}{a^2}$$

Entonces **el conjunto de soluciones de Lasso** es

$$\Big\{(\hat\beta_1, \hat\beta_2) \ :\ \hat\beta_1 + \hat\beta_2 = \hat s,\ \ \hat\beta_1\hat\beta_2 \ge 0\Big\}$$

- Si $\hat s > 0$: el segmento $\hat\beta_1 = t\,\hat s,\ \hat\beta_2 = (1-t)\,\hat s$ con $t \in [0,1]$, que va de $(\hat s, 0)$ a $(0, \hat s)$.
- Si $\hat s < 0$: el mismo segmento en el tercer cuadrante.
- Si $\lambda \ge 4|ay_1|$: $\hat s = 0$ y la solución es **única**, $\hat\beta = (0, 0)$.

Salvo en ese último caso, hay **infinitas soluciones**: todas dan las mismas predicciones y la misma penalización, así que Lasso no puede distinguirlas. Tanto "todo el peso en $x_1$" como "todo en $x_2$" como "mitad y mitad" son óptimos; cuál devuelve un algoritmo depende del algoritmo y de su inicialización (coordinate descent, por ejemplo, suele poner todo en la primera variable que visita).

**Geometría.** En la forma con restricción $|\beta_1| + |\beta_2| \le t$, las curvas de nivel del RSS son las rectas $\beta_1 + \beta_2 = \text{cte}$ (porque el RSS solo depende de la suma). Esas rectas son **paralelas a los lados del rombo** en el primer y tercer cuadrante. Entonces la curva de nivel no toca al rombo en un vértice, sino que se apoya sobre **un lado entero**: ese lado es el segmento de soluciones. En Ridge la región es un círculo, que no tiene lados planos, y la recta lo toca en un único punto: el de la diagonal $\beta_1 = \beta_2$.

**Conclusión.** Es la versión extrema de la frase del enunciado: Ridge **reparte** el peso entre variables correlacionadas (acá, idénticas) en partes iguales; Lasso es **indiferente** al reparto y puede devolver coeficientes muy distintos para ellas, típicamente quedándose con una y apagando la otra. El problema de Lasso sigue siendo convexo, pero no *estrictamente* convexo, y por eso su conjunto de soluciones puede ser un segmento en vez de un punto. Esta es la motivación de Elastic Net (Zou & Hastie, 2005): agregarle un poco de $L_2$ lo vuelve estrictamente convexo y recupera el reparto parejo.


```python
# Verificación numérica del caso n = p = 2
from sklearn.linear_model import Lasso, Ridge

a, y1 = 1.5, 2.0
X = np.array([[a, a], [-a, -a]])
y = np.array([y1, -y1])
lam = 2.0

def obj_ridge(b):
    return np.sum((y - X @ b) ** 2) + lam * np.sum(b ** 2)

def obj_lasso(b):
    return np.sum((y - X @ b) ** 2) + lam * np.sum(np.abs(b))

# 3.2 — Ridge: sklearn minimiza ||y - Xb||^2 + alpha ||b||^2 -> alpha = lambda
ridge = Ridge(alpha=lam, fit_intercept=False).fit(X, y)
print("Ridge sklearn:          ", ridge.coef_)
print("Fórmula 2 a y1/(4a²+λ): ", 2 * a * y1 / (4 * a**2 + lam))

# 3.4 — Lasso: el objetivo es constante sobre todo el segmento de soluciones
s_hat = np.sign(a * y1) * max(abs(a * y1) - lam / 4, 0) / a**2
print(f"\ns_hat = {s_hat:.4f}")
for t in [0.0, 0.25, 0.5, 1.0]:
    b = np.array([t * s_hat, (1 - t) * s_hat])
    print(f"beta = ({b[0]:.4f}, {b[1]:.4f})  ->  objetivo Lasso = {obj_lasso(b):.6f}")
# Si los signos difieren, la penalización crece y el objetivo empeora
b = np.array([s_hat + 1, -1.0])
print(f"beta = ({b[0]:.4f}, {b[1]:.4f})  ->  objetivo Lasso = {obj_lasso(b):.6f}  (signos opuestos: peor)")

# sklearn minimiza (1/2n)||y - Xb||^2 + alpha ||b||_1; con n = 2 -> alpha = lambda / 4
lasso = Lasso(alpha=lam / 4, fit_intercept=False).fit(X, y)
print("\nLasso sklearn (coordinate descent):", lasso.coef_, "-> eligió una de las infinitas soluciones")
```
