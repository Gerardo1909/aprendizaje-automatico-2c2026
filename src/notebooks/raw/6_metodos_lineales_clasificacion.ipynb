### Métodos Lineales para Clasificación

**¿Qué los hace *lineales*?**

Para todos los modelos de clasificación:

* Nuestro predictor $G(x)$ toma valores en el conjunto de etiquetas $\mathcal{G} = \{1, 2, \ldots, K\}$. 
* El espacio de entrada está dividido en regiones, cada una asociada a una clase.

En estos casos en particular, **las fronteras de decisión son hiperplanos en el espacio de entrada.**

**Existen múltiples métodos que dan lugar a fronteras lineales.**

**¿Se puede usar regresión lineal?**

En el caso multiclase, se puede codificar la variable objetivo usando variables indicadoras (*dummy*). Entonces, si $\mathcal{G}$ tiene $K$ clases, habrá $K$ de estos indicadores $Y_k$, $k = 1, \ldots, K$, con $Y_k = 1$ si $G = k$ y $0$ en caso contrario.  Codificando las $N$ instancias de entrenamiento, obtenemos una *matriz indicadora de respuestas* $\mathbf{Y}$ de tamaño $N \times K$, donde cada fila tiene un único 1 y el resto son 0.

(es decir, los coeficientes para el k-ésimo resultado son simplemente las estimaciones de mínimos cuadrados de la regresión de 
$\mathbf{y}_k$ en $\mathbf{X}$).

Para predecir la clase de una nueva instancia $x$, se calcula el vector de respuestas predichas como 

$$ \hat{f}(x) = (1, x^T) \hat{\mathbf{B}}$$

que es un vector de $K$ valores reales. La predicción de clase es la clase con el valor más alto:

$$ \hat{G}(x) = \arg \max_k \hat{f}_k(x).$$

### Análisis Discriminante Lineal (LDA)

**Teorema de Bayes**

$$ P(G = k | X = x) = \frac{P(X = x | G = k) P(G = k)}{P(X = x)}.$$

Vamos a cambiar un poco la notación:

* $\pi_k$ es la probabilidad a priori de la clase $k$.
    * Debe cumplirse que $\sum_{k=1}^K \pi_k = 1$.
* $f_k(x)$ es la densidad condicional de $X$ dado que $G = k$.

Entonces, Bayes queda así:

$$ P(G = k | X = x) = \frac{f_k(x) \pi_k}{\sum_{m=1}^K f_m(x) \pi_m}.$$

Y es suficiente con aproximar cada elemento de la derecha para obtener una buena aproximación de la izquierda (probabilidad posterior).

### Análisis Discriminante Lineal (LDA)

**Teorema de Bayes**

$$ P(G = k | X = x) = \frac{f_k(x) \pi_k}{\sum_{m=1}^K f_m(x) \pi_m}.$$

Debemos aproximar u obtener cada término. 

* Para $f_k(x)$, vamos a usar la vieja y querida distribución normal multivariada:

$$f_k(x) = \frac{1}{(2 \pi)^{p/2} |\mathbf{\Sigma}_k|^{1/2}} e^{-\frac{1}{2} (x - \mu_k)^T \mathbf{\Sigma}_k^{-1} (x - \mu_k)}$$

En el caso particular de LDA, se asume que todas las clases comparten la misma matriz de covarianza, es decir, $\mathbf{\Sigma}_k = \mathbf{\Sigma}$ para todo $k$.

Con esta simple propuesta, ya podemos obtener fronteras de decisión.

La frontera de decisión entre dos clases $k$ y $l$ es el lugar donde $P(G = k | X = x) = P(G = l | X = x)$. Pero más fácil que trabajar con probabilidades es trabajar con *logaritmos de probabilidades*. Entonces

$$ P(G = k | X = x) = P(G = l | X = x)$$

es equivalente a 

$$ \log P(G = k | X = x) = \log P(G = l | X = x).$$

$$ \log P(G = k | X = x) - \log P(G = l | X = x) = 0$$

$$ \log \frac{P(G = k | X = x)} {P(G = l | X = x)} = 0$$


### Análisis Discriminante Lineal (LDA)

$$ P(G = k | X = x) = \frac{f_k(x) \pi_k}{\sum_{m=1}^K f_m(x) \pi_m}.$$

$$f_k(x) = \frac{1}{(2 \pi)^{p/2} |\mathbf{\Sigma}|^{1/2}} e^{-\frac{1}{2} (x - \mu_k)^T \mathbf{\Sigma}^{-1} (x - \mu_k)}$$

Para la frontera de decisión, tenemos que:

$$ \log \frac{P(G = k | X = x)} {P(G = l | X = x)} = 0$$

Calculemos uno de los términos:

$$ \log P(G = k | X = x) = \log f_k(x) + \log \pi_k - \log \sum_{m=1}^K f_m(x) \pi_m $$

Notar que **el último término no depende de $k$**, por lo que será igual para ambas clases y al restar los logaritmos de las probabilidades posteriores, **se cancela**.

Entonces queda:

$$ \log \frac{P(G = k | X = x)} {P(G = l | X = x)} =  \log f_k(x) + \log \pi_k - \log f_l(x) - \log \pi_l $$

$$ \log \frac{P(G = k | X = x)} {P(G = l | X = x)} =  \log \frac{f_k(x)}{f_l(x)} + \log \frac{\pi_k}{\pi_l} $$


### Análisis Discriminante Lineal (LDA)

$$ \log \frac{P(G = k | X = x)} {P(G = l | X = x)} =  \log \frac{f_k(x)}{f_l(x)} + \log \frac{\pi_k}{\pi_l} $$

Veamos que da el $\log f_k(x)$:

$$ \log f_k(x) = \log (\frac{1}{(2 \pi)^{p/2} |\mathbf{\Sigma}|^{1/2}} e^{-\frac{1}{2} (x - \mu_k)^T \mathbf{\Sigma}^{-1} (x - \mu_k)})$$

$$ \log f_k(x) = -\frac{p}{2} \log(2 \pi) - \frac{1}{2} \log |\mathbf{\Sigma}| - \frac{1}{2} (x - \mu_k)^T \mathbf{\Sigma}^{-1} (x - \mu_k)$$

Nuevamente, los dos primeros términos no dependen de $k$, por lo que se cancelan al restar los logaritmos de las probabilidades posteriores. Notar que esto es así porque **asumimos que todas las clases comparten la misma matriz de covarianza**.

### Análisis Discriminante Lineal (LDA)

$$ \log \frac{P(G = k | X = x)} {P(G = l | X = x)} =  \log \frac{f_k(x)}{f_l(x)} + \log \frac{\pi_k}{\pi_l} $$


$$ \log f_k(x) = -\frac{p}{2} \log(2 \pi) - \frac{1}{2} \log |\mathbf{\Sigma}| - \frac{1}{2} (x - \mu_k)^T \mathbf{\Sigma}^{-1} (x - \mu_k)$$


$$ \log \frac{P(G = k | X = x)} {P(G = l | X = x)} =  - \frac{1}{2} (x - \mu_k)^T \mathbf{\Sigma}^{-1} (x - \mu_k) + \frac{1}{2} (x - \mu_l)^T \mathbf{\Sigma}^{-1} (x - \mu_l) + \log \frac{\pi_k}{\pi_l} $$

que, aplicando la distributiva, queda:

$$ \log \frac{P(G = k | X = x)} {P(G = l | X = x)} =  - \frac{1}{2} (\mu_k + \mu_l)^T \mathbf{\Sigma}^{-1}  (\mu_k - \mu_l) + x^T \mathbf{\Sigma}^{-1} (\mu_k - \mu_l) + \log \frac{\pi_k}{\pi_l} $$

que es lineal en $x$ e, igualando a 0, define un hiperplano en el espacio de entrada.


### Análisis Discriminante Lineal (LDA)

Si en lugar de hacer la resta $\log P(G = k | X = x) - \log P(G = l | X = x)$ nos quedábamos únicamente con aquellos términos que dependen de $k$, obteníamos la función discriminante lineal para la clase $k$:

$$ \delta_k(x) = x^T \mathbf{\Sigma}^{-1} \mu_k - \frac{1}{2} \mu_k^T \mathbf{\Sigma}^{-1} \mu_k + \log \pi_k $$

Que es suficiente para clasificar:
1. Calcular $\delta_k(x)$ para cada clase $k$.
2. Asignar $x$ a la clase con mayor valor de $\delta_k(x)$, $\hat{G}(x) = \arg \max_k \delta_k(x)$.

### Análisis Discriminante Lineal (LDA)

$$ \log \frac{P(G = k | X = x)} {P(G = l | X = x)} =  - \frac{1}{2} (\mu_k + \mu_l)^T \mathbf{\Sigma}^{-1}  (\mu_k - \mu_l) + x^T \mathbf{\Sigma}^{-1} (\mu_k - \mu_l) + \log \frac{\pi_k}{\pi_l} $$

Define una frontera de decisión, pero nos falta obtener los parámetros $\mu_k$, $\mathbf{\Sigma}$ y $\pi_k$ (para todas las clases). Obviamente, debemos obtenerlos a partir de los datos de entrenamiento.

* $\pi_k$ se estima como la proporción de instancias de la clase $k$ en el conjunto de entrenamiento. Es decir, $\hat{\pi}_k = \frac{N_k}{N}$, donde $N_k$ es la cantidad de instancias de la clase $k$ y $N$ es la cantidad total de instancias.

* $\hat{\mu}_k = \frac{1}{N_k} \sum_{g_i = k} x_i$. **Recordar que $x$ es multivariado**.

* $\hat{\mathbf{\Sigma}} = \frac{1}{N - K} \sum_{k=1}^K \sum_{g_i = k} (x_i - \hat{\mu}_k)(x_i - \hat{\mu}_k)^T$.



### Análisis Discriminante Cuadrático (QDA)

En LDA, la función discriminante lineal para la clase $k$ era:

$$ \delta_k(x) = x^T \mathbf{\Sigma}^{-1} \mu_k - \frac{1}{2} \mu_k^T \mathbf{\Sigma}^{-1} \mu_k + \log \pi_k $$

que era suficiente para clasificar.

¿Qué ocurría si no asumíamos que todas las clases compartían la misma matriz de covarianza?

En este caso, algunos términos no se cancelarían en la resta de los logaritmos de las probabilidades posteriores, y la frontera de decisión entre dos clases $k$ y $l$ ya no sería lineal en $x$, sino cuadrática. En particular, la función discriminante para la clase $k$ sería:

$$ \delta_k(x) = -\frac{1}{2} \log |\mathbf{\Sigma}_k| - \frac{1}{2} (x - \mu_k)^T \mathbf{\Sigma}_k^{-1} (x - \mu_k) + \log \pi_k $$


### Análisis Discriminante

¿Qué significa entrenar un modelo de LDA o QDA? Notar que estamos asumiendo una normal multivariada para los datos. Geométricamente, esto significa que cada clase tiene una *nube de puntos* con forma elipsoidal en el espacio de entrada.

Este elipsoide puede estar *centrado* en cualquier punto del espacio (definido por el vector de medias $\mu_k$), y *orientado* en cualquier dirección (definido por la matriz de covarianza $\mathbf{\Sigma}_k$).

¿No sería más conveniente si este elipsoide estuviera *alineado* con (algunos) ejes?

Desde el punto de vista de los cálculos, esto significa que la matriz de covarianza es diagonal (no hay términos de covarianza, sólo varianzas). Y, en ese caso, es mucho más fácil obtener su inversa.


### Análisis Discriminante

Vamos a usar la descomposición espectral de la matriz de covarianza (base de autovectores):

$$ \hat{\mathbf{\Sigma}}_k = \mathbf{U}_k \mathbf{D}_k \mathbf{U}_k^T $$

Como $\hat{\mathbf{\Sigma}}_k$ es simétrica y si, además, asumimos que es definida positiva:
* $\mathbf{U}_k$ es una matriz ortogonal cuyas columnas son los autovectores de $\hat{\mathbf{\Sigma}}_k$,
* $\mathbf{D}_k$ es una matriz diagonal cuyos elementos son los autovalores, que además son positivos.
* $\hat{\mathbf{\Sigma}}_k$ tiene dimensión $p \times p$, por lo que $\mathbf{U}_k$ y $\mathbf{D}_k$ también.


Usando esta información, podemos reescribir

* $ (x - \mu_k)^T \hat{\mathbf{\Sigma}}_k^{-1} (x - \mu_k) = (x - \mu_k)^T \mathbf{U}_k \mathbf{D}_k^{-1} \mathbf{U}_k^T (x - \mu_k) = (\mathbf{U}_k^T (x - \mu_k))^T \mathbf{D}_k^{-1} (\mathbf{U}_k^T (x - \mu_k)) $

* $ \log |\hat{\mathbf{\Sigma}}_k| = \sum_{l=1}^p \log \lambda_{kl} $, donde $\lambda_{kl}$ es el l-ésimo autovalor de $\hat{\mathbf{\Sigma}}_k$ (elemento diagonal de $\mathbf{D}_k$).

¿Cómo se usa en LDA?

Entrenamiento:

1. Estimar la matriz de covarianza común y descomponerla: $\hat{\mathbf{\Sigma}} = \mathbf{U} \mathbf{D} \mathbf{U}^T$.
1. Hacer el cambio de variable $X^* = \mathbf{D}^{-1/2} \mathbf{U}^T X$. Con este cambio de variable, ahora la matriz de covarianza es la identidad. Las medias se transforman igual: $\hat{\mu}_k^* = \mathbf{D}^{-1/2} \mathbf{U}^T \hat{\mu}_k$.

Clasificación de una nueva instancia $x$:

1. Transformarla: $x^* = \mathbf{D}^{-1/2} \mathbf{U}^T x$.
1. Asignarla a la clase con el centroide más cercano en el espacio transformado, corrigiendo por la probabilidad a priori:

$$ \hat{G}(x) = \arg \min_k \left\{ \frac{1}{2} \| x^* - \hat{\mu}_k^* \|^2 - \log \hat{\pi}_k \right\} $$

### Intermezzo: Newton-Raphson

El método de Newton-Raphson es un método iterativo para encontrar raíces de funciones reales. 

Empezamos por una función $f(x)$ y queremos encontrar un valor $x^*$ tal que $f(x^*) = 0$; es decir, su raíz o al menos una de sus raíces.

Para ello, vamos a empezar por un valor inicial $x_0$, y aproximaremos la función $f(x)$ por su aproximación lineal (la recta tangente) en ese punto $x_0$:

$$ y = f(x_0) + f'(x_0)(x - x_0) $$

Luego se obtiene la raíz de esa recta tangente. Se espera que la raíz encontrada sea una mejor aproximación a la raíz de $f(x)$. 

$$x_1 = x_0 - \frac{f(x_0)}{f'(x_0)}$$

Esto se repite hasta converger.

### Intermezzo: Distribución multinomial

Es una generalización de la distribución binomial para más de dos clases. Se utiliza para modelar la probabilidad de que una observación pertenezca a una de varias categorías mutuamente excluyentes. La función de masa de probabilidad para la distribución multinomial es:

$$ P(X_1 = x_1, X_2 = x_2, \ldots, X_k = x_k) = \frac{n!}{x_1! x_2! \ldots x_k!} p_1^{x_1} p_2^{x_2} \ldots p_k^{x_k} $$

donde $n$ es el número total de ensayos, $k$ es el número de categorías, $x_i$ es el número de éxitos en la categoría $i$, y $p_i$ es la probabilidad de éxito en la categoría $i$. ($\sum_{i=1}^k x_i = n$ y $\sum_{i=1}^k p_i = 1$).

### Regresión Logística

La regresión lineal aplicada a problemas de clasificación tenía el problema de que los valores obtenidos no se podían interpretar como probabilidades, ya que podían estar fuera del rango $[0, 1]$.

¿Hay alguna alternativa más adecuada?

Como modelar la probabilidad directamente no es sencillo, vamos a modelar el *log-odds*:

Supongamos que tenemos $K$ clases; entonces, la probabilidad condicional de que, dado un vector de características $x$, la instancia pertenezca a la clase $k$ es $P(G = k | X = x)$. Teniendo en cuenta que $\sum_{k=1}^K P(G = k | X = x) = 1$, podemos definir los *log-odds* en función de una clase de referencia (por ejemplo, la clase $K$):

$$ \log \frac{P(G = k | X = x)}{P(G = K | X = x)} = \beta_{k0} + \beta_k^T x, \quad k = 1, \ldots, K-1 $$



### Regresión Logística

$$ \log \frac{P(G = 1 | X = x)}{P(G = K | X = x)} = \beta_{10} + \beta_1^T x $$

$$ \log \frac{P(G = 2 | X = x)}{P(G = K | X = x)} = \beta_{20} + \beta_2^T x $$

$$ \vdots $$

$$ \log \frac{P(G = K-1 | X = x)}{P(G = K | X = x)} = \beta_{(K-1)0} + \beta_{K-1}^T x $$



### Regresión Logística

Despejando $P(G = k | X = x)$ y $P(G = K | X = x)$, obtenemos:

$$ P(G = k | X = x) = \frac{e^{\beta_{k0} + \beta_k^T x}}{1 + \sum_{l=1}^{K-1} e^{\beta_{l0} + \beta_l^T x}}, \quad k = 1, \ldots, K-1 $$

$$ P(G = K | X = x) = \frac{1}{1 + \sum_{l=1}^{K-1} e^{\beta_{l0} + \beta_l^T x}} $$

Notar que las probabilidades dependen de todo el conjunto de parámetros $\theta = \{\beta_{10}, \beta_1^T, \ldots, \beta_{(K-1)0}, \beta_{K-1}^T\}$.

Para simplificar la notación, vamos a llamar a $P(G = k | X = x)$ como $p_k(x;\theta)$.

### Regresión Logística

Ejemplifiquemos el caso más sencillo, el binario ($K=2$).

$$ \log \frac{P(G = 1 | X = x)}{P(G = 2 | X = x)} = \log \frac{p_1(x;\theta)}{p_2(x;\theta)} = \beta_{10} + \beta_1^T x $$

Entonces, despejando, queda:

$$ p_1(x;\theta) = \frac{e^{\beta_{10} + \beta_1^T x}}{1 + e^{\beta_{10} + \beta_1^T x}} = \frac{1}{1 + e^{-(\beta_{10} + \beta_1^T x)}} $$

y 

$$ p_2(x;\theta) = 1 - p_1(x;\theta) = \frac{1}{1 + e^{\beta_{10} + \beta_1^T x}} $$

que sólo tiene involucrados el conjunto de parámetros asociado a una única función lineal, $\theta = (\beta_{10}, \beta_1)$.

### Regresión Logística

#### Entrenamiento

La regresión logística se entrena típicamente mediante máxima verosimilitud. ¿Cuál es la distribución más apropiada, considerando que la variable objetivo es categórica?

¡La distribución multinomial! Ya que nos permite modelar la probabilidad de que una observación pertenezca a una de varias categorías mutuamente excluyentes.

La verosimilitud es

$$ \mathcal{L}(\theta) = \prod_{i=1}^N P(G = g_i | X = x_i) = \prod_{i=1}^N p_{g_i}(x_i;\theta) $$

pero es más fácil trabajar con el logaritmo de la verosimilitud:

$$ \ell(\theta) =  \sum_{i=1}^N \log p_{g_i}(x_i;\theta) $$

### Regresión Logística

#### Entrenamiento

Nuevamente, es más fácil trabajar con el caso binario ($K=2$). Luego mencionaremos algunas estrategias para el caso general.
Vamos a codificar la variable objetivo como $y_i = 1$ si $g_i = 1$ y $y_i = 0$ si $g_i = 2$. Como son solamente dos casos, solamente es necesario modelar $p_1(x;\theta)$, ya que $p_2(x;\theta) = 1 - p_1(x;\theta)$. Para simplificar la notación, vamos a llamar $p(x_i;\beta) = p_1(x_i;\theta)$, donde $\beta = \{\beta_{10}, \beta_1\}$. Entonces, el logaritmo de la verosimilitud queda:

$$ \ell(\beta) =  \sum_{i=1}^N \{ y_i \log p(x_i;\beta) + (1 - y_i) \log (1 - p(x_i;\beta)) \} $$

Notar que de esta forma se cubren los dos casos posibles para $y_i$.

Teniendo en cuenta que $ p(x_i;\beta) = \frac{e^{\beta^T x_i}}{1 + e^{(\beta^T x_i)}} $ (se le agrega un 1 a $x_i$ para el término independiente), y que $1 - p(x_i;\beta) = \frac{1}{1 + e^{\beta^T x_i}}$, queda:

$$ \ell(\beta) =  \sum_{i=1}^N \{ y_i \log \frac{e^{\beta^T x_i}}{1 + e^{\beta^T x_i}} + (1 - y_i) \log \frac{1}{1 + e^{\beta^T x_i}} \} $$

$$ \ell(\beta) =  \sum_{i=1}^N \{ y_i \beta^T x_i - \log (1 + e^{\beta^T x_i}) \} $$

### Regresión Logística

#### Entrenamiento

Entonces, la función a maximizar es

$$ \ell(\beta) =  \sum_{i=1}^N \{ y_i \beta^T x_i - \log (1 + e^{\beta^T x_i}) \} $$

que es una función de $\beta$, por lo que va de $\mathbb{R}^{p+1} \to \mathbb{R}$.

Para maximizar tenemos que obtener su derivada e igualarla a cero:

$$ \frac{\partial \ell(\beta)}{\partial \beta} = \sum_{i=1}^N x_i (y_i - p(x_i;\beta)) = 0$$

¿Cuántas ecuaciones son? ¿Qué es la derivada en este caso?

### Regresión Logística

#### Entrenamiento

$$ \frac{\partial \ell(\beta)}{\partial \beta} = \sum_{i=1}^N x_i (y_i - p(x_i;\beta)) = 0$$

Recordar: queremos encontrar los valores de $\beta$ que maximizan la función de log-verosimilitud (y cumplen con estas ecuaciones).

Son, en total, $p+1$ ecuaciones no-lineales (una por cada componente de $\beta$). Veamos algunas de estas ecuaciones más en detalle.

* Como $x_i$ tiene un 1 en la primera posición (término independiente), la primera ecuación es:

$$ \sum_{i=1}^N (y_i - p(x_i;\beta)) = 0 \quad \Rightarrow \quad \sum_{i=1}^N y_i = \sum_{i=1}^N p(x_i;\beta) $$

que se cumple cuando la suma esperada de elementos de la clase 1 es igual a la suma observada.

* Para las otras ecuaciones, $j = 1, \ldots, p$, tenemos que:

$$ \sum_{i=1}^N x_{ij} (y_i - p(x_i;\beta)) = 0 \quad \Rightarrow \quad \sum_{i=1}^N x_{ij} y_i = \sum_{i=1}^N x_{ij} p(x_i;\beta) $$

$$ \sum_{i=1}^N x_{ij} y_i = \sum_{i=1}^N x_{ij} \frac{e^{\beta^T x_i}}{1 + e^{\beta^T x_i}} $$

### Regresión Logística

#### Entrenamiento

$$ \frac{\partial \ell(\beta)}{\partial \beta} = \sum_{i=1}^N x_i (y_i - p(x_i;\beta)) = 0$$

Para resolver este sistema no lineal de ecuaciones, podemos usar el método de Newton-Raphson. Como es el caso en el que buscamos extremos de una función, necesitamos la derivada segunda, es decir, la matriz Hessiana en el caso multidimensional:

$$ \mathbf{H} =   \frac{\partial^2 \ell(\beta)}{\partial \beta \partial \beta^T} =  \sum_{i=1}^N - x_i x_i^T p(x_i;\beta) (1 - p(x_i;\beta)) $$

**Importante**: ¿Qué es $x_i x_i^T$? Es el producto exterior, que da como resultado una matriz.

Por lo tanto, la actualización de Newton-Raphson es:

$$ \beta^{(new)} = \beta^{(old)} - (\frac{\partial^2 \ell(\beta)}{\partial \beta \partial \beta^T})_{\beta = \beta^{(old)}}^{-1} \frac{\partial \ell(\beta)}{\partial \beta}_{\beta = \beta^{(old)}} $$

Con el ojo ya entrenado, ¿dónde pueden aparecer problemas?

### Regresión Logística

#### Entrenamiento

Repasemos:

* El gradiente es $\frac{\partial \ell(\beta)}{\partial \beta} = \sum_{i=1}^N x_i (y_i - p(x_i;\beta))$.
* El Hessiano es $\mathbf{H} =   \frac{\partial^2 \ell(\beta)}{\partial \beta \partial \beta^T} = \sum_{i=1}^N -x_i x_i^T p(x_i;\beta) (1 - p(x_i;\beta)) $.
* La actualización de Newton-Raphson es $ \beta^{(new)} = \beta^{(old)} - (\frac{\partial^2 \ell(\beta)}{\partial \beta \partial \beta^T})_{\beta = \beta^{(old)}}^{-1} \frac{\partial \ell(\beta)}{\partial \beta}_{\beta = \beta^{(old)}} $

A esta altura de la materia, ya sabemos que es mejor trabajar con matrices y vectores. Entonces, vamos a definir:

* $\mathbf{y} = (y_1, y_2, \ldots, y_N)^T$
* $\mathbf{X}$ es la matriz de diseño, de tamaño $N \times (p+1)$
* $\mathbf{p} = (p(x_1;\beta), p(x_2;\beta), \ldots, p(x_N;\beta))^T$, el vector de probabilidades predichas para todas las instancias.
* $\mathbf{W}$ es una matriz diagonal de tamaño $N \times N$, con entradas $W_{ii} = p(x_i;\beta)(1 - p(x_i;\beta))$

### Regresión Logística

#### Entrenamiento

Usando la notación de la diapositiva anterior, podemos reescribir el gradiente y el Hessiano como:

* Gradiente: $\frac{\partial \ell(\beta)}{\partial \beta} = \mathbf{X}^T (\mathbf{y} - \mathbf{p})$
* Hessiano: $\mathbf{H} = - \mathbf{X}^T \mathbf{W} \mathbf{X}$

Por lo tanto, la actualización de Newton-Raphson queda en forma matricial como 

$$ \beta^{(new)} = \beta^{(old)} + (\mathbf{X}^T \mathbf{W} \mathbf{X})^{-1} \mathbf{X}^T (\mathbf{y} - \mathbf{p}) $$




### Regresión Logística

#### Entrenamiento

***Iteratively Reweighted Least Squares* (IRLS)**

Podemos complicar las cosas un poco más.

$$ \beta^{(new)} = \beta^{(old)} + (\mathbf{X}^T \mathbf{W} \mathbf{X})^{-1} \mathbf{X}^T (\mathbf{y} - \mathbf{p}) $$

Multiplicando por la identidad $  (\mathbf{X}^T \mathbf{W} \mathbf{X})^{-1} \mathbf{X}^T \mathbf{W} \mathbf{X} $, la actualización queda

$$ \beta^{(new)} = (\mathbf{X}^T \mathbf{W} \mathbf{X})^{-1} \mathbf{X}^T \mathbf{W} \mathbf{X} \beta^{(old)} + (\mathbf{X}^T \mathbf{W} \mathbf{X})^{-1} \mathbf{X}^T (\mathbf{y} - \mathbf{p}) $$

$$ \beta^{(new)} = (\mathbf{X}^T \mathbf{W} \mathbf{X})^{-1} \mathbf{X}^T \mathbf{W} (\mathbf{X} \beta^{(old)} + \mathbf{W}^{-1} (\mathbf{y} - \mathbf{p})) $$

Si llamamos $\mathbf{z} = \mathbf{X} \beta^{(old)} + \mathbf{W}^{-1} (\mathbf{y} - \mathbf{p})$, queda

$$ \beta^{(new)} = (\mathbf{X}^T \mathbf{W} \mathbf{X})^{-1} \mathbf{X}^T \mathbf{W} \mathbf{z} $$

Que, ¡es la solución de un problema de mínimos cuadrados ponderados!

$$ \beta^{(new)} = \arg \min_\beta (\mathbf{z} - \mathbf{X} \beta)^T \mathbf{W} (\mathbf{z} - \mathbf{X} \beta) $$

Es decir, en cada iteración, se resuelve un problema de mínimos cuadrados ponderados, donde los pesos dependen de la iteración anterior.

¿Cuál es su ventaja?

* Hereda todos los métodos de regresión lineal (ponderada). Esto facilita usar rutinas numéricamente estables ya existentes y con buena convergencia.
* **Extensión a modelos lineales generalizados (GLM)**. Esta formulación es general: cualquier GLM con función de enlace canónica puede resolverse con IRLS, usando una definición adecuada de $\mathbf{z}$ y $\mathbf{W}$.


### Regresión Logística

#### Regularización

Se maximiza la función de log-verosimilitud penalizada:

**L1 (Lasso)**

$$ \max_{\beta_0, \beta} \left\{ \sum_{i=1}^N \ [y_i (\beta_0 + \beta^T x_i) - \log(1 + e^{\beta_0 + \beta^T x_i})] - \lambda \sum_{j=1}^p |\beta_j| \right\} $$

**L2 (Ridge)**

$$ \max_{\beta_0, \beta} \left\{ \sum_{i=1}^N \ [y_i (\beta_0 + \beta^T x_i) - \log(1 + e^{\beta_0 + \beta^T x_i})] - \lambda \sum_{j=1}^p \beta_j^2 \right\} $$

donde, como ocurría en regresión lineal, el término de penalización no incluye al término independiente $\beta_0$ y los predictores deben estar estandarizados.
